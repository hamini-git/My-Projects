```python
df['Date'] = pd.to_datetime(df['Date'])
df.info()
```

```python
df[df.isnull().any(axis=1)]
```

```python
df.dropna(inplace=True,axis=0)
df
```

```python
df.duplicated().sum()
df.drop_duplicates(inplace=True)
df.reset_index(drop=True, inplace=True)
df
```

```python
df['Total'] = df['Quantity'] * df['Price']
mismatched_rows = df[df['Total_Amount'] != df['Total']]
print("Mismatched rows:")
print(mismatched_rows)
df['Total_Amount'] = df['Total']
df.drop(columns=['Total'], inplace=True)
```

```python
total_missing = df.isna().sum()*100/len(df)
print('Percentage Missing Value %')
total_missing
```

```python
# Numerical column - Data Visualization
fig, axes = plt.subplots(3, 2, figsize=(10, 8))  # define plot sizes
axes = axes.flatten()                            # flatten axes
for i, col in enumerate(numerical_columns_list):  # loop through columns
  axes[i].hist(df[col], bins=20, edgecolor='black')     # define axes for histogram
  axes[i].set_title(f'Histogram of {col}')          # define title for histogram
  axes[i].set_xlabel(col)                           # define x-axis label
  axes[i].set_ylabel('Frequency')                   # define y-axis label
for j in range(i + 1, len(axes)):                   # Hide Unused axes
    fig.delaxes(axes[j])                            # delete unused axes

plt.tight_layout()                                  # Prevent Overlapping and adjust layout
plt.show()                                          # plot visualization charts for numeric data
```

```python
# Numerical column - Data Visualization
fig, axes = plt.subplots(3, 2, figsize=(10, 8))  # define plot sizes
axes = axes.flatten()
for i, col in enumerate(categorical_columns_list):  # loop through columns
  axes[i].hist(df[col], bins=20, edgecolor='black')     # define axes for histogram
  axes[i].set_title(f'Histogram of {col}')          # define title for histogram
  axes[i].set_xlabel(col)                           # define x-axis label
  axes[i].set_ylabel('Frequency')                   # define y-axis label
for j in range(i + 1, len(axes)):                   # Hide Unused axes
    fig.delaxes(axes[j])                            # delete unused axes

plt.tight_layout()                                  # Prevent Overlapping and adjust layout
plt.show()                                          # plot visualization charts for numeric data
```

```python
sales_by_date = df.groupby('Date')['Total_Amount'].sum().reset_index()
sales_by_date
```

```python
df['Month'] = df['Date'].dt.to_period('M')
```

```python
df['Month'] = df['Month'].dt.to_timestamp()
```

```python
monthly_sales = df.groupby('Month')['Total_Amount'].sum().reset_index()
```

```python
plt.figure(figsize=(10, 5))
sns.lineplot(data=monthly_sales, x='Month', y='Total_Amount', marker='o')
plt.title('Monthly Wise Sales')
plt.xlabel('Month')
plt.ylabel('Total Sales Amount')
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
product_sales = df.groupby('Product')['Total_Amount'].sum().reset_index()
product_sales = product_sales.sort_values(by='Total_Amount', ascending=False)
product_sales
```

```python
df1=df.drop(columns=['Transaction_ID']) # drop unnecessary columns
df1
```

```python
from sklearn.preprocessing import LabelEncoder# Initialize LabelEncoder
label_encoder = LabelEncoder()

# Apply LabelEncoder to each categorical column
for column in df1.columns:
    if df1[column].dtype == 'object':  # Check if column is categorical
        df1[column] = label_encoder.fit_transform(df1[column])
        # convert object type to numeric type
# Now all columns should be numeric
print(df1.dtypes)  # Check data types to ensure all are numeric
df1  # print data set after datatype conversion
```

```python
df1.corr()
```

```python
correlation = df1.corr(numeric_only=True)
plt.figure(figsize=(10, 6))
sns.heatmap(correlation, annot=True, fmt=".2f", cmap='YlGnBu', linewidths=0.5)
plt.title('Correlation Heatmap')
plt.tight_layout()
plt.show()
```

```python
region_month = df.groupby(['Region', 'Month'])['Total_Amount'].sum().reset_index()
region_month
```

```python
pivot_region = region_month.pivot(index='Region', columns='Month', values='Total_Amount').fillna(0)
plt.figure(figsize=(12, 6))
sns.heatmap(pivot_region, annot=True, fmt=".1f", cmap="YlGnBu")
plt.title("Region-wise Sales by Month")
plt.xlabel("Month")
plt.ylabel("Region")
plt.tight_layout()
plt.show()
```
